# 04 — KoVAE Comparison Plots with Center-Crop Reconstruction



Main changes:

```text
1. Keeps your two-method KoVAE structure:
   - rollout_v1
   - posterior_bank_v2

2. Uses future-proof foldering:
   figures/comparison/kovae/<method_name>/
   results/comparison/kovae/<method_name>/

   Later TimeVAE can use:
   figures/comparison/timevae/<method_name>/
   results/comparison/timevae/<method_name>/

3. Adds center-crop reconstruction:
   - Real spaced timeline: overlap-add using real metadata
   - Real center-crop timeline: middle stride chunk from each real window
   - Synthetic center-crop timeline: middle stride chunk from each synthetic window

4. Uses common y-axis per three-way comparison plot.

5. Supports multiple real subjects and multiple synthetic subjects.

6. Duration matching is optional and disabled by default.
```

Why center-crop is useful:

Generated windows are independent synthetic windows. They are not guaranteed to form a true continuous recording. Center-crop reconstruction avoids pretending that overlapping synthetic windows are perfectly continuous.


````python

# ============================================================
# 04_comparison_kovae_center_crop_dual_methods.py
#
# Updated Notebook 04 for KoVAE comparison.
#
# Main idea:
#   - Real spaced timeline:
#       overlap-add reconstruction using real metadata and sample positions.
#
#   - Real center-crop timeline:
#       compressed reconstruction from the middle stride-sized chunk of each
#       real window.
#
#   - Synthetic center-crop timeline:
#       compressed reconstruction from the middle stride-sized chunk of each
#       synthetic window.
#
# Why:
#   Synthetic windows are generated window-by-window. They are not guaranteed
#   to be a true continuous overlapping recording. Center-crop visualization
#   is therefore safer and easier to defend in the report.
#
# Foldering:
#   figures/comparison/kovae/<method>/
#   results/comparison/kovae/<method>/
#
# Later TimeVAE can use:
#   figures/comparison/timevae/<method>/
#   results/comparison/timevae/<method>/
# ============================================================

from __future__ import annotations

from pathlib import Path
from typing import Dict, List, Optional, Tuple
import json
import re
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# ============================================================
# Config
# ============================================================

COMPARISON_CONFIG = {
    "project_root": "/home/iailab42/khans1/projects/ir",

    # Important for clean future foldering.
    # Later for TimeVAE use model_family="timevae".
    "model_family": "kovae",

    "real_dir": "data/processed/native_rates",
    "synthetic_base_dir": "data/synthetic_subjects/kovae",

    # Base folders. The code will automatically create:
    # figures/comparison/kovae/<method>/
    # results/comparison/kovae/<method>/
    "figures_base_dir": "figures/comparison",
    "results_base_dir": "results/comparison",

    "methods_to_compare": ["rollout_v1", "posterior_bank_v2"],

    # Use several subjects because one subject can be misleading.
    "real_subjects_to_compare": ["S1", "S2", "S3"],

    # Synthetic selection:
    #   "first_n"  -> first N synthetic subjects
    #   "specific" -> use specific_synthetic_subjects
    "synthetic_subject_selection": "first_n",
    "num_synthetic_subjects_to_compare": 3,
    "specific_synthetic_subjects": [],

    # Pairing:
    #   "same_index" -> S1 vs synthetic_1, S2 vs synthetic_2, ...
    #   "all_pairs"  -> every selected real subject vs every selected synthetic subject
    "pairing_mode": "same_index",

    # Duration matching is optional.
    # With 3000 windows, synthetic subject duration is already close to real.
    # Keep False if you want full available timelines.
    "match_duration_for_three_way": False,

    # If match_duration_for_three_way=True:
    #   None -> use min available duration across the three panels
    #   number -> use that many seconds, capped by available duration
    "comparison_duration_sec": None,
    "comparison_start_sec": 0.0,

    # Plot all points or decimate.
    # Use 1 for full quality. Use 2,4,8 if plotting is slow.
    "plot_step": 1,

    # Common y-axis makes amplitude comparison fair.
    "use_common_y_axis": True,
    "use_robust_y_axis": True,
    "robust_y_percentiles": [0.5, 99.5],

    "show_activity_background": True,
    "save_plot": True,

    # Saving continuous arrays can produce many files.
    # Figures and CSV summaries are saved either way.
    "save_continuous_arrays": False,
}


# ============================================================
# Branch configs
# ============================================================

BRANCH_CONFIGS = {
    "ACC": {
        "real_filename": "all_X_acc_32hz.npy",
        "syn_filename": "generated_subjects_X_acc_32hz.npy",
        "hz": 32,
        "window_len": 256,
        "shift_len": 64,
        "num_channels": 3,
        "channel_names": ["ACC_x", "ACC_y", "ACC_z"],
        "start_col": "acc_start_sample_32hz",
        "end_col": "acc_end_sample_32hz",
    },
    "BVP": {
        "real_filename": "all_X_bvp_64hz.npy",
        "syn_filename": "generated_subjects_X_bvp_64hz.npy",
        "hz": 64,
        "window_len": 512,
        "shift_len": 128,
        "num_channels": 1,
        "channel_names": ["BVP"],
        "start_col": "bvp_start_sample_64hz",
        "end_col": "bvp_end_sample_64hz",
    },
    "SLOW": {
        "real_filename": "all_X_slow_4hz.npy",
        "syn_filename": "generated_subjects_X_slow_4hz.npy",
        "hz": 4,
        "window_len": 32,
        "shift_len": 8,
        "num_channels": 2,
        "channel_names": ["EDA", "TEMP"],
        "start_col": "slow_start_sample_4hz",
        "end_col": "slow_end_sample_4hz",
    },
}


# ============================================================
# Basic helpers
# ============================================================

def sanitize_name(value: str) -> str:
    value = str(value)
    value = re.sub(r"[^A-Za-z0-9_.-]+", "_", value)
    return value.strip("_")


def subject_sort_key(value):
    text = str(value)

    if text.startswith("S") and text[1:].isdigit():
        return (0, int(text[1:]))

    numbers = re.findall(r"\d+", text)

    if numbers:
        return (1, int(numbers[-1]), text)

    return (2, text)


def require_file(path: Path) -> Path:
    path = Path(path)

    if not path.exists():
        raise FileNotFoundError(f"Missing file: {path}")

    return path


def make_dirs(*dirs: Path) -> None:
    for directory in dirs:
        directory.mkdir(parents=True, exist_ok=True)


def save_json(data: Dict, path: Path) -> None:
    path.write_text(json.dumps(data, indent=2), encoding="utf-8")


def get_base_paths(config: Dict) -> Dict[str, Path]:
    root = Path(config["project_root"])
    model_family = config["model_family"]

    return {
        "root": root,
        "real_dir": root / config["real_dir"],
        "synthetic_base_dir": root / config["synthetic_base_dir"],
        "figures_model_dir": root / config["figures_base_dir"] / model_family,
        "results_model_dir": root / config["results_base_dir"] / model_family,
        "configs_dir": root / "configs",
    }


def get_method_output_dirs(base_paths: Dict[str, Path], method_name: str) -> Dict[str, Path]:
    results_dir = base_paths["results_model_dir"] / method_name
    figures_dir = base_paths["figures_model_dir"] / method_name

    return {
        "results_dir": results_dir,
        "figures_dir": figures_dir,
        "arrays_dir": results_dir / "continuous_arrays",
        "summary_dir": results_dir / "summaries",
    }


def check_array_shape(X: np.ndarray, branch_name: str, cfg: Dict, source_name: str) -> None:
    if X.ndim != 3:
        raise ValueError(f"{source_name} {branch_name}: expected [N,T,C], got {X.shape}")

    if X.shape[1] != int(cfg["window_len"]):
        raise ValueError(
            f"{source_name} {branch_name}: expected T={cfg['window_len']}, got {X.shape[1]}"
        )

    if X.shape[2] != int(cfg["num_channels"]):
        raise ValueError(
            f"{source_name} {branch_name}: expected C={cfg['num_channels']}, got {X.shape[2]}"
        )


# ============================================================
# Loading
# ============================================================

def load_real_common(real_dir: Path) -> Tuple[np.ndarray, np.ndarray, pd.DataFrame]:
    y_path = real_dir / "all_y.npy"
    subject_path = real_dir / "all_subject.npy"
    metadata_path = real_dir / "all_metadata.csv"

    require_file(y_path)
    require_file(subject_path)

    y = np.load(y_path).astype(np.int64)
    subjects = np.load(subject_path, allow_pickle=True).astype(str)

    if metadata_path.exists():
        metadata = pd.read_csv(metadata_path)
    else:
        warnings.warn(f"Missing real metadata: {metadata_path}. Fallback timing will be used.")
        metadata = pd.DataFrame(index=np.arange(len(y)))

    if len(y) != len(subjects):
        raise ValueError(f"Real y/subject mismatch: {len(y)} vs {len(subjects)}")

    if len(metadata) != len(y):
        raise ValueError(f"Real metadata/y mismatch: {len(metadata)} vs {len(y)}")

    metadata = metadata.copy()
    metadata["subject"] = subjects
    metadata["activity_label"] = y
    metadata["array_index"] = np.arange(len(y), dtype=np.int64)

    return y, subjects, metadata


def load_synthetic_common(synthetic_dir: Path) -> Tuple[np.ndarray, np.ndarray, pd.DataFrame]:
    y_path = synthetic_dir / "generated_subjects_all_y.npy"
    subject_path = synthetic_dir / "generated_subjects_all_subject.npy"
    metadata_path = synthetic_dir / "generated_subjects_metadata.csv"

    require_file(y_path)
    require_file(subject_path)
    require_file(metadata_path)

    y = np.load(y_path).astype(np.int64)
    subjects = np.load(subject_path, allow_pickle=True).astype(str)
    metadata = pd.read_csv(metadata_path)

    if len(y) != len(subjects):
        raise ValueError(f"Synthetic y/subject mismatch: {len(y)} vs {len(subjects)}")

    if len(metadata) != len(y):
        raise ValueError(f"Synthetic metadata/y mismatch: {len(metadata)} vs {len(y)}")

    metadata = metadata.copy()
    metadata["synthetic_subject"] = subjects
    metadata["activity_label"] = y
    metadata["array_index"] = np.arange(len(y), dtype=np.int64)

    return y, subjects, metadata


def load_branch_arrays(real_dir: Path, synthetic_dir: Path) -> Tuple[Dict[str, np.ndarray], Dict[str, np.ndarray]]:
    real_X = {}
    syn_X = {}

    for branch_name, cfg in BRANCH_CONFIGS.items():
        real_path = real_dir / cfg["real_filename"]
        syn_path = synthetic_dir / cfg["syn_filename"]

        require_file(real_path)
        require_file(syn_path)

        X_real = np.load(real_path).astype(np.float32)
        X_syn = np.load(syn_path).astype(np.float32)

        check_array_shape(X_real, branch_name, cfg, "Real")
        check_array_shape(X_syn, branch_name, cfg, "Synthetic")

        real_X[branch_name] = X_real
        syn_X[branch_name] = X_syn

    return real_X, syn_X


def load_all_data_for_method(
    real_dir: Path,
    synthetic_dir: Path,
) -> Tuple[
    Dict[str, np.ndarray],
    np.ndarray,
    np.ndarray,
    pd.DataFrame,
    Dict[str, np.ndarray],
    np.ndarray,
    np.ndarray,
    pd.DataFrame,
]:
    real_y, real_subjects, real_metadata = load_real_common(real_dir)
    syn_y, syn_subjects, syn_metadata = load_synthetic_common(synthetic_dir)
    real_X, syn_X = load_branch_arrays(real_dir, synthetic_dir)

    for branch_name in BRANCH_CONFIGS:
        if len(real_X[branch_name]) != len(real_y):
            raise ValueError(
                f"Real {branch_name}/y mismatch: {len(real_X[branch_name])} vs {len(real_y)}"
            )

        if len(syn_X[branch_name]) != len(syn_y):
            raise ValueError(
                f"Synthetic {branch_name}/y mismatch: {len(syn_X[branch_name])} vs {len(syn_y)}"
            )

    return real_X, real_y, real_subjects, real_metadata, syn_X, syn_y, syn_subjects, syn_metadata


# ============================================================
# Subject selection
# ============================================================

def get_selected_synthetic_subjects(
    syn_subjects: np.ndarray,
    config: Dict,
) -> List[str]:
    unique_subjects = sorted(
        np.unique(syn_subjects.astype(str)).tolist(),
        key=subject_sort_key,
    )

    mode = config["synthetic_subject_selection"]

    if mode == "first_n":
        n = int(config["num_synthetic_subjects_to_compare"])
        return unique_subjects[:n]

    if mode == "specific":
        selected = [str(x) for x in config["specific_synthetic_subjects"]]
        missing = sorted(set(selected) - set(unique_subjects))

        if missing:
            raise ValueError(f"Specific synthetic subjects not found: {missing}")

        return selected

    raise ValueError("synthetic_subject_selection must be 'first_n' or 'specific'.")


def build_subject_pairs(
    real_subjects: List[str],
    synthetic_subjects: List[str],
    config: Dict,
) -> List[Tuple[str, str]]:
    mode = config["pairing_mode"]

    if mode == "same_index":
        n = min(len(real_subjects), len(synthetic_subjects))
        return [(real_subjects[i], synthetic_subjects[i]) for i in range(n)]

    if mode == "all_pairs":
        return [(r, s) for r in real_subjects for s in synthetic_subjects]

    raise ValueError("pairing_mode must be 'same_index' or 'all_pairs'.")


# ============================================================
# Timing and reconstruction helpers
# ============================================================

def ensure_timing_columns(
    metadata: pd.DataFrame,
    subject_col: str,
    start_col: str,
    end_col: str,
    window_len: int,
    shift_len: int,
) -> pd.DataFrame:
    metadata = metadata.copy().reset_index(drop=True)

    if start_col in metadata.columns and end_col in metadata.columns:
        return metadata

    warnings.warn(
        f"Missing timing columns {start_col}/{end_col}. "
        "Using fallback timing based on window order inside each subject."
    )

    if subject_col not in metadata.columns:
        metadata[subject_col] = "unknown_subject"

    within_subject_index = metadata.groupby(subject_col).cumcount().to_numpy(dtype=np.int64)

    metadata[start_col] = within_subject_index * int(shift_len)
    metadata[end_col] = metadata[start_col] + int(window_len)

    return metadata


def print_consecutiveness_report(
    metadata: pd.DataFrame,
    start_col: str,
    expected_shift: int,
    title: str,
) -> Dict[str, float]:
    starts = metadata[start_col].to_numpy(dtype=np.int64)

    if len(starts) < 2:
        report = {
            "num_windows": int(len(starts)),
            "num_transitions": 0,
            "num_expected_shift_transitions": 0,
            "expected_shift_ratio": np.nan,
            "median_start_diff": np.nan,
            "max_start_diff": np.nan,
        }
        print(title, report)
        return report

    starts = np.sort(starts)
    diffs = np.diff(starts)

    expected_count = int(np.sum(diffs == int(expected_shift)))

    report = {
        "num_windows": int(len(starts)),
        "num_transitions": int(len(diffs)),
        "num_expected_shift_transitions": expected_count,
        "expected_shift_ratio": float(expected_count / max(1, len(diffs))),
        "median_start_diff": float(np.median(diffs)),
        "max_start_diff": float(np.max(diffs)),
    }

    print(title)
    print(
        f"windows={report['num_windows']} | "
        f"expected_shift_ratio={report['expected_shift_ratio']:.3f} | "
        f"median_diff={report['median_start_diff']:.1f} | "
        f"max_diff={report['max_start_diff']:.1f}"
    )

    return report


def reconstruct_spaced_overlap_add(
    windows: np.ndarray,
    y: np.ndarray,
    metadata: pd.DataFrame,
    start_col: str,
    end_col: str,
    num_activity_classes: int = 8,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Reconstruct a spaced timeline using start/end sample positions.

    Returns:
        continuous: [T_total, C]
        sample_labels: [T_total]
        coverage: [T_total]
    """

    if len(windows) == 0:
        raise ValueError("Cannot reconstruct empty windows.")

    starts = metadata[start_col].to_numpy(dtype=np.int64)
    ends = metadata[end_col].to_numpy(dtype=np.int64)

    min_start = int(np.min(starts))
    starts = starts - min_start
    ends = ends - min_start

    order = np.argsort(starts)
    windows = windows[order]
    y = y[order]
    starts = starts[order]
    ends = ends[order]

    T_total = int(np.max(ends))
    _, window_len, num_channels = windows.shape

    continuous_sum = np.zeros((T_total, num_channels), dtype=np.float64)
    weight_sum = np.zeros((T_total, 1), dtype=np.float64)
    label_scores = np.zeros((T_total, num_activity_classes + 1), dtype=np.float64)

    weights = np.hanning(window_len).astype(np.float64)

    if np.all(weights == 0):
        weights = np.ones(window_len, dtype=np.float64)

    weights = np.maximum(weights, 1e-4)

    for i in range(len(windows)):
        start = int(starts[i])
        end = int(ends[i])

        if end <= start:
            continue

        segment_len = min(window_len, end - start)
        end = start + segment_len

        w = weights[:segment_len, None]
        continuous_sum[start:end] += windows[i, :segment_len, :] * w
        weight_sum[start:end] += w

        label = int(y[i])

        if 0 <= label <= num_activity_classes:
            label_scores[start:end, label] += weights[:segment_len]

    coverage = weight_sum[:, 0]
    valid = coverage > 0

    continuous = np.zeros_like(continuous_sum, dtype=np.float32)
    continuous[valid] = (continuous_sum[valid] / weight_sum[valid]).astype(np.float32)

    sample_labels = np.argmax(label_scores, axis=1).astype(np.int64)
    sample_labels[~valid] = 0

    return continuous.astype(np.float32), sample_labels, coverage.astype(np.float32)


def reconstruct_windows_center_crop(
    windows: np.ndarray,
    y: np.ndarray,
    shift_len: int,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Reconstruct a compressed timeline by taking only the middle stride-sized
    segment from every window.

    This is safer for synthetic windows because generated windows are not
    guaranteed to form a true overlapping continuous recording.
    """

    if len(windows) == 0:
        raise ValueError("Cannot reconstruct empty windows.")

    _, window_len, _ = windows.shape
    shift_len = int(shift_len)

    if shift_len > window_len:
        raise ValueError(f"shift_len={shift_len} cannot be larger than window_len={window_len}")

    start = (window_len - shift_len) // 2
    end = start + shift_len

    cropped = windows[:, start:end, :].astype(np.float32)
    continuous = cropped.reshape(len(windows) * shift_len, windows.shape[2])

    sample_labels = np.repeat(y.astype(np.int64), shift_len)

    return continuous.astype(np.float32), sample_labels.astype(np.int64)


def crop_by_seconds(
    continuous: np.ndarray,
    labels: np.ndarray,
    target_hz: int,
    start_sec: float,
    duration_sec: Optional[float],
) -> Tuple[np.ndarray, np.ndarray]:
    if duration_sec is None:
        return continuous, labels

    start_idx = int(round(float(start_sec) * float(target_hz)))
    duration_samples = int(round(float(duration_sec) * float(target_hz)))
    end_idx = start_idx + duration_samples

    start_idx = max(0, start_idx)
    end_idx = min(len(continuous), end_idx)

    if end_idx <= start_idx:
        raise ValueError(
            f"Invalid crop range. start_idx={start_idx}, end_idx={end_idx}, len={len(continuous)}"
        )

    return continuous[start_idx:end_idx], labels[start_idx:end_idx]


def decide_comparison_duration(
    real_spaced: np.ndarray,
    real_center: np.ndarray,
    syn_center: np.ndarray,
    target_hz: int,
    config: Dict,
) -> Optional[float]:
    if not bool(config["match_duration_for_three_way"]):
        return None

    available = min(
        len(real_spaced) / float(target_hz),
        len(real_center) / float(target_hz),
        len(syn_center) / float(target_hz),
    )

    requested = config["comparison_duration_sec"]

    if requested is None:
        return float(available)

    return float(min(float(requested), available))


# ============================================================
# Plot helpers
# ============================================================

def decimate_for_plot(
    time_sec: np.ndarray,
    signal: np.ndarray,
    labels: np.ndarray,
    step: int,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    step = int(step)

    if step <= 1:
        return time_sec, signal, labels

    return time_sec[::step], signal[::step], labels[::step]


def find_label_segments(labels: np.ndarray) -> List[Tuple[int, int, int]]:
    if len(labels) == 0:
        return []

    labels = labels.astype(np.int64)
    segments = []

    start = 0
    current = int(labels[0])

    for i in range(1, len(labels)):
        label = int(labels[i])

        if label != current:
            segments.append((start, i, current))
            start = i
            current = label

    segments.append((start, len(labels), current))

    return segments


def add_activity_background(
    ax,
    time_sec: np.ndarray,
    labels: np.ndarray,
    min_segment_sec_for_text: float = 20.0,
) -> None:
    if len(time_sec) == 0 or len(labels) == 0:
        return

    segments = find_label_segments(labels)

    if len(time_sec) > 1:
        dt = float(np.median(np.diff(time_sec)))
    else:
        dt = 1.0

    for start_idx, end_idx, label in segments:
        if int(label) <= 0:
            continue

        start_time = float(time_sec[start_idx])
        end_time = float(time_sec[min(end_idx - 1, len(time_sec) - 1)] + dt)

        ax.axvspan(start_time, end_time, alpha=0.08)

        duration = end_time - start_time

        if duration >= min_segment_sec_for_text:
            mid_time = (start_time + end_time) / 2.0
            y_min, y_max = ax.get_ylim()
            text_y = y_min + 0.06 * (y_max - y_min)
            ax.text(
                mid_time,
                text_y,
                f"act {int(label)}",
                ha="center",
                va="bottom",
                fontsize=8,
                alpha=0.8,
            )


def compute_common_ylim(
    arrays: List[np.ndarray],
    channel_idx: int,
    use_robust: bool,
    percentiles: List[float],
) -> Optional[Tuple[float, float]]:
    values = []

    for arr in arrays:
        if arr is None or len(arr) == 0:
            continue

        values.append(arr[:, channel_idx].reshape(-1))

    if not values:
        return None

    values = np.concatenate(values)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return None

    if use_robust:
        low, high = np.percentile(values, percentiles)
    else:
        low, high = np.min(values), np.max(values)

    low = float(low)
    high = float(high)

    if low == high:
        pad = 1.0
    else:
        pad = 0.08 * (high - low)

    return low - pad, high + pad


def plot_three_way_channel(
    real_spaced: np.ndarray,
    real_spaced_labels: np.ndarray,
    real_center: np.ndarray,
    real_center_labels: np.ndarray,
    syn_center: np.ndarray,
    syn_center_labels: np.ndarray,
    channel_idx: int,
    channel_name: str,
    branch_name: str,
    real_subject: str,
    synthetic_subject: str,
    method_name: str,
    target_hz: int,
    save_path: Path,
    config: Dict,
) -> None:
    duration_sec = decide_comparison_duration(
        real_spaced=real_spaced,
        real_center=real_center,
        syn_center=syn_center,
        target_hz=target_hz,
        config=config,
    )

    start_sec = float(config["comparison_start_sec"])

    real_spaced, real_spaced_labels = crop_by_seconds(
        real_spaced,
        real_spaced_labels,
        target_hz,
        start_sec=start_sec,
        duration_sec=duration_sec,
    )

    real_center, real_center_labels = crop_by_seconds(
        real_center,
        real_center_labels,
        target_hz,
        start_sec=start_sec,
        duration_sec=duration_sec,
    )

    syn_center, syn_center_labels = crop_by_seconds(
        syn_center,
        syn_center_labels,
        target_hz,
        start_sec=start_sec,
        duration_sec=duration_sec,
    )

    real_spaced_time = np.arange(len(real_spaced)) / float(target_hz)
    real_center_time = np.arange(len(real_center)) / float(target_hz)
    syn_center_time = np.arange(len(syn_center)) / float(target_hz)

    step = int(config["plot_step"])

    real_spaced_time, real_spaced_plot, real_spaced_labels_plot = decimate_for_plot(
        real_spaced_time,
        real_spaced,
        real_spaced_labels,
        step,
    )

    real_center_time, real_center_plot, real_center_labels_plot = decimate_for_plot(
        real_center_time,
        real_center,
        real_center_labels,
        step,
    )

    syn_center_time, syn_center_plot, syn_center_labels_plot = decimate_for_plot(
        syn_center_time,
        syn_center,
        syn_center_labels,
        step,
    )

    ylim = None

    if bool(config["use_common_y_axis"]):
        ylim = compute_common_ylim(
            arrays=[real_spaced_plot, real_center_plot, syn_center_plot],
            channel_idx=channel_idx,
            use_robust=bool(config["use_robust_y_axis"]),
            percentiles=list(config["robust_y_percentiles"]),
        )

    fig, axes = plt.subplots(3, 1, figsize=(18, 9), sharex=False)

    plot_items = [
        (
            axes[0],
            real_spaced_time,
            real_spaced_plot[:, channel_idx],
            real_spaced_labels_plot,
            f"Real spaced timeline | {real_subject} | {channel_name}",
        ),
        (
            axes[1],
            real_center_time,
            real_center_plot[:, channel_idx],
            real_center_labels_plot,
            f"Real center-crop compressed | {real_subject} | {channel_name}",
        ),
        (
            axes[2],
            syn_center_time,
            syn_center_plot[:, channel_idx],
            syn_center_labels_plot,
            f"Synthetic center-crop | {method_name} | {synthetic_subject} | {channel_name}",
        ),
    ]

    for ax, time_sec, signal, labels, title in plot_items:
        ax.plot(time_sec, signal, linewidth=0.8)

        if ylim is not None:
            ax.set_ylim(*ylim)

        if bool(config["show_activity_background"]):
            add_activity_background(ax, time_sec, labels)

        ax.set_title(title)
        ax.set_ylabel(channel_name)
        ax.grid(alpha=0.2)

    axes[-1].set_xlabel("Time in seconds")

    crop_note = ""

    if duration_sec is not None:
        crop_note = f" | duration matched to {duration_sec:.1f}s"

    fig.suptitle(
        f"{config['model_family']} comparison | method={method_name} | "
        f"real={real_subject} vs synthetic={synthetic_subject} | branch={branch_name}{crop_note}",
        fontsize=12,
    )

    fig.tight_layout(rect=[0, 0, 1, 0.96])

    if bool(config["save_plot"]):
        save_path.parent.mkdir(parents=True, exist_ok=True)
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
        print("Saved:", save_path)

    plt.show()
    plt.close(fig)


# ============================================================
# Pair processing
# ============================================================

def save_continuous_npz(
    path: Path,
    real_spaced: np.ndarray,
    real_spaced_labels: np.ndarray,
    real_center: np.ndarray,
    real_center_labels: np.ndarray,
    syn_center: np.ndarray,
    syn_center_labels: np.ndarray,
) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)

    np.savez_compressed(
        path,
        real_spaced=real_spaced,
        real_spaced_labels=real_spaced_labels,
        real_center=real_center,
        real_center_labels=real_center_labels,
        syn_center=syn_center,
        syn_center_labels=syn_center_labels,
    )

    print("Saved:", path)


def process_one_pair_branch(
    real_X: Dict[str, np.ndarray],
    real_y: np.ndarray,
    real_metadata: pd.DataFrame,
    syn_X: Dict[str, np.ndarray],
    syn_y: np.ndarray,
    syn_metadata: pd.DataFrame,
    real_subject: str,
    synthetic_subject: str,
    branch_name: str,
    method_name: str,
    output_dirs: Dict[str, Path],
    config: Dict,
) -> List[Dict]:
    cfg = BRANCH_CONFIGS[branch_name]

    real_mask = real_metadata["subject"].astype(str).to_numpy() == str(real_subject)
    syn_mask = syn_metadata["synthetic_subject"].astype(str).to_numpy() == str(synthetic_subject)

    if not np.any(real_mask):
        raise ValueError(f"No real windows found for subject {real_subject}")

    if not np.any(syn_mask):
        raise ValueError(f"No synthetic windows found for subject {synthetic_subject}")

    real_windows = real_X[branch_name][real_mask]
    real_labels = real_y[real_mask]
    real_meta_sub = real_metadata.loc[real_mask].copy().reset_index(drop=True)

    syn_windows = syn_X[branch_name][syn_mask]
    syn_labels = syn_y[syn_mask]
    syn_meta_sub = syn_metadata.loc[syn_mask].copy().reset_index(drop=True)

    real_meta_sub = ensure_timing_columns(
        metadata=real_meta_sub,
        subject_col="subject",
        start_col=cfg["start_col"],
        end_col=cfg["end_col"],
        window_len=int(cfg["window_len"]),
        shift_len=int(cfg["shift_len"]),
    )

    syn_meta_sub = ensure_timing_columns(
        metadata=syn_meta_sub,
        subject_col="synthetic_subject",
        start_col=cfg["start_col"],
        end_col=cfg["end_col"],
        window_len=int(cfg["window_len"]),
        shift_len=int(cfg["shift_len"]),
    )

    real_consecutive_report = print_consecutiveness_report(
        metadata=real_meta_sub,
        start_col=cfg["start_col"],
        expected_shift=int(cfg["shift_len"]),
        title=f"Real consecutiveness | {real_subject} | {branch_name}",
    )

    syn_consecutive_report = print_consecutiveness_report(
        metadata=syn_meta_sub,
        start_col=cfg["start_col"],
        expected_shift=int(cfg["shift_len"]),
        title=f"Synthetic consecutiveness | {synthetic_subject} | {branch_name}",
    )

    real_spaced, real_spaced_labels, real_coverage = reconstruct_spaced_overlap_add(
        windows=real_windows,
        y=real_labels,
        metadata=real_meta_sub,
        start_col=cfg["start_col"],
        end_col=cfg["end_col"],
    )

    real_center, real_center_labels = reconstruct_windows_center_crop(
        windows=real_windows,
        y=real_labels,
        shift_len=int(cfg["shift_len"]),
    )

    syn_center, syn_center_labels = reconstruct_windows_center_crop(
        windows=syn_windows,
        y=syn_labels,
        shift_len=int(cfg["shift_len"]),
    )

    pair_name = f"{sanitize_name(real_subject)}_vs_{sanitize_name(synthetic_subject)}"
    pair_fig_dir = output_dirs["figures_dir"] / pair_name
    pair_array_dir = output_dirs["arrays_dir"] / pair_name

    if bool(config["save_continuous_arrays"]):
        save_continuous_npz(
            path=pair_array_dir / f"{branch_name}_continuous_reconstructions.npz",
            real_spaced=real_spaced,
            real_spaced_labels=real_spaced_labels,
            real_center=real_center,
            real_center_labels=real_center_labels,
            syn_center=syn_center,
            syn_center_labels=syn_center_labels,
        )

    rows = []

    duration_sec = decide_comparison_duration(
        real_spaced=real_spaced,
        real_center=real_center,
        syn_center=syn_center,
        target_hz=int(cfg["hz"]),
        config=config,
    )

    for channel_idx, channel_name in enumerate(cfg["channel_names"]):
        fig_path = pair_fig_dir / f"{branch_name}_{channel_name}_three_way_center_crop.png"

        plot_three_way_channel(
            real_spaced=real_spaced,
            real_spaced_labels=real_spaced_labels,
            real_center=real_center,
            real_center_labels=real_center_labels,
            syn_center=syn_center,
            syn_center_labels=syn_center_labels,
            channel_idx=channel_idx,
            channel_name=channel_name,
            branch_name=branch_name,
            real_subject=real_subject,
            synthetic_subject=synthetic_subject,
            method_name=method_name,
            target_hz=int(cfg["hz"]),
            save_path=fig_path,
            config=config,
        )

        rows.append(
            {
                "model_family": config["model_family"],
                "method": method_name,
                "real_subject": real_subject,
                "synthetic_subject": synthetic_subject,
                "branch": branch_name,
                "channel": channel_name,
                "hz": int(cfg["hz"]),
                "window_len": int(cfg["window_len"]),
                "shift_len": int(cfg["shift_len"]),
                "real_num_windows": int(len(real_windows)),
                "synthetic_num_windows": int(len(syn_windows)),
                "real_spaced_duration_sec": float(len(real_spaced) / float(cfg["hz"])),
                "real_center_duration_sec": float(len(real_center) / float(cfg["hz"])),
                "synthetic_center_duration_sec": float(len(syn_center) / float(cfg["hz"])),
                "comparison_duration_sec": None if duration_sec is None else float(duration_sec),
                "real_expected_shift_ratio": real_consecutive_report["expected_shift_ratio"],
                "synthetic_expected_shift_ratio": syn_consecutive_report["expected_shift_ratio"],
                "figure_path": str(fig_path),
            }
        )

    return rows


def process_one_method(method_name: str, config: Dict) -> pd.DataFrame:
    base_paths = get_base_paths(config)
    synthetic_dir = base_paths["synthetic_base_dir"] / method_name
    output_dirs = get_method_output_dirs(base_paths, method_name)

    make_dirs(
        output_dirs["results_dir"],
        output_dirs["figures_dir"],
        output_dirs["arrays_dir"],
        output_dirs["summary_dir"],
    )

    print("\n" + "#" * 100)
    print(f"COMPARISON METHOD: {method_name}")
    print("#" * 100)
    print("Real dir:     ", base_paths["real_dir"])
    print("Synthetic dir:", synthetic_dir)
    print("Figures dir:  ", output_dirs["figures_dir"])
    print("Results dir:  ", output_dirs["results_dir"])

    (
        real_X,
        real_y,
        real_subjects,
        real_metadata,
        syn_X,
        syn_y,
        syn_subjects,
        syn_metadata,
    ) = load_all_data_for_method(
        real_dir=base_paths["real_dir"],
        synthetic_dir=synthetic_dir,
    )

    available_real = set(real_subjects.astype(str))
    selected_real = [str(s) for s in config["real_subjects_to_compare"]]

    missing_real = sorted(set(selected_real) - available_real, key=subject_sort_key)

    if missing_real:
        raise ValueError(f"Requested real subjects not found: {missing_real}")

    selected_syn = get_selected_synthetic_subjects(
        syn_subjects=syn_subjects,
        config=config,
    )

    pairs = build_subject_pairs(
        real_subjects=selected_real,
        synthetic_subjects=selected_syn,
        config=config,
    )

    print("\nSelected real subjects:", selected_real)
    print("Selected synthetic subjects:", selected_syn)
    print("Pairing mode:", config["pairing_mode"])
    print("Pairs:")
    for r, s in pairs:
        print("  ", r, "vs", s)

    all_rows = []

    for real_subject, synthetic_subject in pairs:
        print("\n" + "=" * 100)
        print(f"Pair: {real_subject} vs {synthetic_subject}")
        print("=" * 100)

        for branch_name in BRANCH_CONFIGS:
            rows = process_one_pair_branch(
                real_X=real_X,
                real_y=real_y,
                real_metadata=real_metadata,
                syn_X=syn_X,
                syn_y=syn_y,
                syn_metadata=syn_metadata,
                real_subject=real_subject,
                synthetic_subject=synthetic_subject,
                branch_name=branch_name,
                method_name=method_name,
                output_dirs=output_dirs,
                config=config,
            )
            all_rows.extend(rows)

    summary_df = pd.DataFrame(all_rows)

    summary_path = output_dirs["summary_dir"] / "comparison_plot_summary.csv"
    summary_df.to_csv(summary_path, index=False)
    print("\nSaved:", summary_path)

    method_summary = {
        "model_family": config["model_family"],
        "method": method_name,
        "synthetic_dir": str(synthetic_dir),
        "figures_dir": str(output_dirs["figures_dir"]),
        "results_dir": str(output_dirs["results_dir"]),
        "summary_csv": str(summary_path),
        "num_rows": int(len(summary_df)),
        "selected_real_subjects": selected_real,
        "selected_synthetic_subjects": selected_syn,
        "pairing_mode": config["pairing_mode"],
        "uses_center_crop_reconstruction": True,
        "uses_common_y_axis": bool(config["use_common_y_axis"]),
        "match_duration_for_three_way": bool(config["match_duration_for_three_way"]),
    }

    save_json(method_summary, output_dirs["summary_dir"] / "method_comparison_summary.json")

    return summary_df


def main(config: Dict = COMPARISON_CONFIG) -> Dict[str, object]:
    base_paths = get_base_paths(config)

    make_dirs(
        base_paths["figures_model_dir"],
        base_paths["results_model_dir"],
        base_paths["configs_dir"],
    )

    config_path = base_paths["configs_dir"] / f"comparison_{config['model_family']}_center_crop_config.json"
    save_json(config, config_path)

    print("=" * 100)
    print("Notebook 04: comparison with center-crop reconstruction")
    print("=" * 100)
    print("Model family:", config["model_family"])
    print("Methods:", config["methods_to_compare"])
    print("Config saved to:", config_path)
    print("Figures model dir:", base_paths["figures_model_dir"])
    print("Results model dir:", base_paths["results_model_dir"])

    method_summaries = {}

    for method_name in config["methods_to_compare"]:
        method_summaries[method_name] = process_one_method(
            method_name=method_name,
            config=config,
        )

    combined_df = pd.concat(method_summaries.values(), ignore_index=True)

    combined_csv = base_paths["results_model_dir"] / "combined_comparison_plot_summary.csv"
    combined_df.to_csv(combined_csv, index=False)

    combined_summary = {
        "model_family": config["model_family"],
        "methods_to_compare": config["methods_to_compare"],
        "figures_model_dir": str(base_paths["figures_model_dir"]),
        "results_model_dir": str(base_paths["results_model_dir"]),
        "combined_summary_csv": str(combined_csv),
        "num_plot_rows": int(len(combined_df)),
        "foldering_note": {
            "kovae": "figures/comparison/kovae/<method>/ and results/comparison/kovae/<method>/",
            "future_timevae": "figures/comparison/timevae/<method>/ and results/comparison/timevae/<method>/",
        },
    }

    combined_json = base_paths["results_model_dir"] / "combined_comparison_summary.json"
    save_json(combined_summary, combined_json)

    print("\n" + "#" * 100)
    print("Comparison completed.")
    print("#" * 100)
    print("Combined CSV:", combined_csv)
    print("Combined JSON:", combined_json)

    return {
        "combined_summary": combined_summary,
        "combined_df": combined_df,
        "method_summaries": method_summaries,
    }


if __name__ == "__main__":
    outputs = main(COMPARISON_CONFIG)
````
